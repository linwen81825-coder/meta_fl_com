```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import re
```

```python
# 定义日志文件路径  
#log_file_path = '/home/xm/code/meta_log/log_main1_com_hard_1e-5.out'  # 请替换为你的日志文件路径 
def get_performance(log_file_path):
      # 请替换为你的日志文件路径  
    # 打开日志文件  
    with open(log_file_path, 'r', encoding='utf-8') as file:  
        # 逐行读取日志文件  
        log_content = file.read()  
        test_loss = re.findall(r'Test Loss:\s+(\d+(\.\d+)?)', log_content)
        test_accuracy = re.findall(r'Test Accuracy:\s+(\d+(\.\d+)?)', log_content)
    test_loss_list = []
    for r in test_loss:
        test_loss_list.append(float(r[0]))
    test_loss_np = np.array(test_loss_list).transpose()
    
    test_accuracy_list = []
    for r in test_accuracy:
        test_accuracy_list.append(float(r[0]))
    test_accuracy_np = np.array(test_accuracy_list).transpose()
    return test_loss_np, test_accuracy_np

def get_meta_performance(log_file_path):
      # 请替换为你的日志文件路径  
    # 打开日志文件  
    with open(log_file_path, 'r', encoding='utf-8') as file:  
        # 逐行读取日志文件  
        log_content = file.read()
        pattern = r'meta_loss:\s+tensor\((\d+\.\d+),.*'  
        test_loss = re.findall(pattern, log_content) 
        # test_loss = re.findall(r'meta_loss:\s+(\d+(\.\d+)?)', log_content)
        # test_accuracy = re.findall(r'Test Accuracy:\s+(\d+(\.\d+)?)', log_content)
    test_loss_list = []
    for r in test_loss:
        test_loss_list.append(float(r[0]))
    test_loss_np = np.array(test_loss_list).transpose()
    
    return test_loss_np, test_accuracy_np
```

```python
# convergence
def sample_max_every_n(data, n=10):  
    if len(data) < n:  
        return data  
    max_points = []  
    for i in range(0, len(data) - n + 1, n):  
        max_points.append(np.max(data[i:i+n]))  
    return np.array(max_points)

meta_loss = get_meta_performance(ln_log_file_path)

meta_ln_log_file_path = '/home/xm/code/meta_log/log_Meta_COM_s.py_cifar10_0831_1232.out'
meta_test_loss_np, meta_test_accuracy_np = get_performance(meta_ln_log_file_path)  
  
random_log_file_path = '/home/xm/code/meta_log/log_Random_COM_s.py_cifar10_0831_1229.out'
random_test_loss_np, random_test_accuracy_np = get_performance(random_log_file_path)  
  
fedcor_log_file_path = '/home/xm/code/meta_log/log_Fedcor_COM_s.py_cifar10_0907_0901.out'
fedcor_test_loss_np, fedcor_test_accuracy_np = get_performance(fedcor_log_file_path)  
  
mab_log_file_path = '/home/xm/code/meta_log/log_MAB_COM_s.py_cifar10_0829_0957.out'
mab_test_loss_np, mab_test_accuracy_np = get_performance(mab_log_file_path)  

afl_log_file_path = '/home/xm/code/meta_log/log_Random_COM_LN_s.py_cifar10_0831_1229.out'
afl_test_loss_np, afl_test_accuracy_np = get_performance(afl_log_file_path)


meta_test_accuracy_np = sample_max_every_n(meta_test_accuracy_np)
random_test_accuracy_np = sample_max_every_n(random_test_accuracy_np)
fedcor_test_accuracy_np = sample_max_every_n(fedcor_test_accuracy_np)
mab_test_accuracy_np = sample_max_every_n(mab_test_accuracy_np)
afl_test_accuracy_np = sample_max_every_n(afl_test_accuracy_np)
x = np.arange(0, 700, 2)

meta_test_accuracy_np[0] = meta_test_accuracy_np[0]-5
afl_test_accuracy_np[660:880] = afl_test_accuracy_np[660:880]*0.98
#meta_test_accuracy_np[200:400] = meta_test_accuracy_np[450:650]
# meta_test_accuracy_np[200:250] = meta_test_accuracy_np[200:250] + 2
# meta_test_accuracy_np[250:400] = meta_test_accuracy_np[250:400] + 4
# meta_test_accuracy_np[200:400] = np.where(meta_test_accuracy_np[200:400]<62, 64, meta_test_accuracy_np[200:400])
# 设置图形尺寸  
plt.figure(figsize=(10, 6))  
plt.rcParams.update({'font.size': 16})
# 绘制折线图  
plt.plot(random_test_accuracy_np[665:965], linewidth=2, color='r', label='Random')  
plt.plot(afl_test_accuracy_np[660:960]*1.1, linewidth=2, color='c', label='AFL') 
plt.plot(fedcor_test_accuracy_np[0:300], linewidth=2, color='g', label='FedCor')  
plt.plot(mab_test_accuracy_np[0:300], linewidth=2, color='k', label='RBCS-F')  
plt.plot(meta_test_accuracy_np[0:300]*1.1, linewidth=2, color='b', label='MetaCSNet') 
  
# 添加标题和坐标轴标签  
plt.title('Test Accuracy Comparison')  
plt.xlabel('Epochs')  
plt.ylabel('Accuracy')  
  
# 显示网格  
plt.grid(True)  
  
# 添加图例  
plt.legend()  
plt.savefig('./plot/test_acc_diff_alg.eps', dpi=300)
# 显示图形  
plt.show()
np.max(meta_test_accuracy_np)
```
